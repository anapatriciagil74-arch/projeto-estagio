```python
import pandas as pd

df = pd.read_csv("adults.csv")
```

```python
df.columns = df.columns.str.strip()
```

```python
df = df.map(lambda x: x.strip() if isinstance(x, str) else x)
```

```python
df.replace("?", pd.NA, inplace=True)
```

```python
print(df["income"].unique())
```

```python
df.to_csv("adults_data.csv", index=False)
```
