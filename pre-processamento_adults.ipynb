```python
import pandas as pd
import numpy as np

# Carregar o CSV com todos os campos como string para evitar problemas de tipo
df = pd.read_csv('adults.csv', dtype=str)

# Limpar espaços nos nomes das colunas e substituir hífens por underscores
# (o Vertex AI não aceita hífens nos nomes das colunas)
df.columns = [c.strip().replace('-', '_') for c in df.columns]

# Limpar espaços nos valores e substituir '?' por NaN
for col in df.columns:
df[col] = df[col].str.strip().replace({'?': 'Unknown', 'nan': 'Unknown', '': 'Unknown'})

# Remover linhas com valores em falta
df = df.replace('Unknown', np.nan).dropna().reset_index(drop=True)

# Verificar resultado
print('Shape:', df.shape) # (30162, 15)
print('Nulos:', df.isna().sum().sum()) # 0
print('Target:', df['income'].unique()) # ['<=50K', '>50K']

# Guardar CSV limpo
df.to_csv('adults_data_clean.csv', index=False)
```
