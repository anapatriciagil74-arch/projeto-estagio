```python
import pandas as pd
import numpy as np

# Carregar CSV com todos os campos como string
df = pd.read_csv('adults.csv', dtype=str)

# Limpar nomes das colunas
df.columns = [c.strip().replace('-', '_') for c in df.columns]

# Limpar espaços e substituir valores desconhecidos por NaN
for col in df.columns:
    df[col] = (
        df[col]
        .str.strip()
        .replace({
            '?': np.nan,
            'Unknown': np.nan,
            'unknown': np.nan,
            'nan': np.nan,
            'NaN': np.nan,
            '': np.nan
        })
    )

# Remover linhas com valores em falta
df = df.dropna().reset_index(drop=True)

# Verificar resultado
print('Shape:', df.shape)
print('Nulos:', df.isna().sum().sum())
print('Target:', df['income'].unique())

# Guardar CSV limpo
df.to_csv('adults_data_clean.csv', index=False)
```
